### Al abrir el archivo csv vemos que contiene tildes y mas caracteres, hay que setear un encoding a corde, y luego sacar las columnas vacias

```python
import pandas as pd

dataset_netflix = pd.read_csv("netflix_titles.csv", encoding="latin-1")

# Sacar columnas no deseadas
columnas_deseadas = ["show_id","type","title","director","cast","country","date_added","release_year","rating","duration","listed_in","description"]

dataset_netflix = dataset_netflix[columnas_deseadas]
```

### ¿Cuántos directores distintos hay?

```python
cantidad_directores = dataset_netflix["director"].nunique()
print("cantidad de directores")
print(cantidad_directores)
```

### ¿Quién es el director con más películas?

```python
director = dataset_netflix["director"].value_counts().idxmax()
print(director)
```

### ¿Cuántas series hay? ¿Y cuántos programas de TV?

```python
cantidad_seriespeliculas = dataset_netflix["type"].value_counts()
print(cantidad_seriespeliculas)

# otra forma

cantidad_series = (dataset_netflix["type"] == "Movie").sum()
print("La cantidad de series totales es", cantidad_series)

cantidad_tv = (dataset_netflix["type"] == "TV Show").sum()
print("La cantidad de series totales es", cantidad_tv)
```

### ¿Cuál es la película más vieja del catálogo? ¿Y la serie?

```python
pelicula = dataset_netflix[dataset_netflix["type"] == "Movie"]
ano_minimo = pelicula['release_year'].min()

pelicula_vieja = pelicula[pelicula["release_year"] == ano_minimo]["title"]

print("Pelicula mas vieja:")
print(pelicula_vieja)
print("Año: ", ano_minimo)
```

```python
pelicula = dataset_netflix[dataset_netflix["type"] == "TV Show"]
ano_minimo = pelicula['release_year'].min()

serie_vieja = pelicula[pelicula["release_year"] == ano_minimo]["title"]

print("Serie mas vieja:")
print(serie_vieja)
print("Año: ", ano_minimo)
```

### Dado un país, determinar cuántas películas de ese país hay en la base de datos.


```python
pais = input().title()
dataset_pelicula = dataset_netflix[dataset_netflix["type"] == "Movie"]

cantidad_pais = dataset_pelicula[dataset_pelicula["country"] == pais]

print("La cantidad de peliculas que hay en el pais es: ",len(cantidad_pais))
print("La cantidad de peliculas que hay en el pais es: ",cantidad_pais.shape[0])
```
