```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Cargar y estandarizar
df = pd.read_csv('ObesityDataSet_raw_and_data_sinthetic.csv')
num_cols = ['Age', 'Height', 'Weight', 'FCVC', 'NCP', 'CH2O', 'FAF', 'TUE']
X_scaled = StandardScaler().fit_transform(df[num_cols])

# Aplicar PCA completo
pca = PCA().fit(X_scaled)
var_acumulada = np.cumsum(pca.explained_variance_ratio_) * 100

# Determinar componentes necesarias para 95%
n_95 = np.argmax(var_acumulada >= 95) + 1
print(f"Componentes necesarias para >= 95% varianza: {n_95} ({var_acumulada[n_95-1]:.2f}%)")
```
